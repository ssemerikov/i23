2. Реалізація косинусної подібності:


```python
def cosine_similarity(u, v):
    return np.dot(u,v) / (np.linalg.norm(u) * np.linalg.norm(v))
```

3. Реалізація евклідової відстані:


```python
def euclidean_distance(u, v):
    return np.linalg.norm(u - v)
```

4. Функція для пошуку аналогій:


```python
def find_analogy(w1, w2, w3, model, cosine_similarity):

    # Відфільтровуємо тільки ті слова, які є в моделі
    for w in (w1, w2, w3):
        if w not in model:
            return None

    vec = model[w3] - model[w1] + model[w2]

    # Шукаємо слово, найближче до отриманого вектора за косинусною подібністю
    similarities = model.cosine_similarities(vec, model.vectors)
    most_similar_index = similarities.argmax()
    return model.index_to_key[most_similar_index]
```

5. Тестування на всьому наборі столиць з підрахунком точності:


```python
def get_country(city1, country1, city2, word_embeddings, cosine_similarity=cosine_similarity):

    # Зберігаємо city1, country1 та city2 у множині group
    group = {city1, country1, city2}

    # Отримуємо векторне представлення city1
    city1_emb = word_embeddings[city1]

    # Отримуємо векторне представлення country1
    country1_emb = word_embeddings[country1]

    # Отримуємо векторне представлення city2
    city2_emb = word_embeddings[city2]

    # Обчислюємо вектор для країни (поєднання векторів country1, city1 і city2)
    # Пам’ятаємо: King - Man + Woman = Queen
    vec = city2_emb - city1_emb + country1_emb

    # Ініціалізуємо схожість значенням -1 (воно буде замінено на значення ближче до +1)
    similarity = -1

    # Ініціалізуємо змінну country як порожній рядок
    country = ''

    # Перебираємо всі слова у словнику векторів
    for word in word_embeddings.index_to_key:

        # Спочатку перевіряємо, чи слово не належить до 'group'
        if word not in group:

            # Отримуємо векторне представлення слова
            word_emb = word_embeddings[word]

            # Обчислюємо косинусну схожість між вектором країни та словом
            cur_similarity = cosine_similarity(vec, word_emb)

            # Якщо поточна схожість краща за попередню найкращу схожість...
            if cur_similarity > similarity:

                # Оновлюємо схожість до нового, кращого значення
                similarity = cur_similarity

                # Зберігаємо країну як кортеж із назвою слова та схожістю
                country = (word, similarity)

    return country
```

```python
def get_accuracy(word_embeddings, data, get_country=get_country):
    # Ініціалізуємо кількість правильних відповідей як нуль
    num_correct = 0

    # Проходимо по рядках датафрейму
    for i, row in data.iterrows():

        # Отримуємо country1 (країна 1)
        country1 = row.iloc[0]

        # Отримуємо city1 (місто 1)
        city1 = row.iloc[1]

        # Отримуємо country2 (країна 2)
        country2 = row.iloc[2]

        # Отримуємо city2 (місто 2)
        city2 = row.iloc[3]

        print(i)  # Виводимо номер рядка
        print(row)  # Виводимо рядок
        # Використовуємо get_country для визначення передбачуваної країни 2
        predicted_country2, _ = get_country(city1, country1, city2, word_embeddings)
        print(predicted_country2)  # Виводимо передбачувану країну
        print("")  # Порожній рядок для відступу

        # Якщо передбачена країна 2 збігається з фактичною країною 2...
        if predicted_country2 == country2:
            # Збільшуємо кількість правильних відповідей на 1
            num_correct += 1

    # Отримуємо кількість рядків у датафреймі (довжина датафрейму)
    m = len(data)

    # Обчислюємо точність, ділячи кількість правильних відповідей на кількість рядків
    accuracy = num_correct / m

    return accuracy


accuracy = get_accuracy(model, data)
print(f"Точність становить {accuracy:.2f}")
```

6. Візуалізація векторів обраних слів за допомогою PCA:


```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

def visualize(words, model):
    # Відфільтровуємо слова, які присутні в словнику моделі
    valid_words = [w for w in words if w in model.key_to_index]

    # Створюємо вектори слів тільки для допустимих слів
    word_vectors = np.array([model[w] for w in valid_words])

    pca = PCA(n_components=2)
    components = pca.fit_transform(word_vectors)

    plt.scatter(components[:, 0], components[:, 1])
    for i, word in enumerate(valid_words):
        plt.annotate(word, xy=(components[i, 0], components[i, 1]))

    plt.show()

```

```python
words = ["Україна", "Київ", "Польща", "Варшава", "Німеччина", "Берлін", "Росія", "Москва"]
visualize(words, model)
```
